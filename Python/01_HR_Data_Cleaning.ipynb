```python
import pandas as pd
import numpy as np
```

## Loading Dataset and Having A copy 

```python
# Load dataset
data = pd.read_csv(r"C:\Users\sujit\OneDrive\Creative Cloud Files\Desktop\Project 1- HR Workforce Analytics & Employee Attrition Analysis\data\Raw\WA_Fn-UseC_-HR-Employee-Attrition.csv")
hr = data.copy()
```

## Checking The Dataset

```python
# Checking Top 5 Rows
print("Top 5 Rows : ")
print(hr.head())
```

```python
# Data Overview
print("Dataset Shape :", hr.shape)

print("\nDataset Type :")
print(hr.dtypes)

print("\nColumns Name : ")
print(hr.columns.tolist())
```

```python
# Check unique employee IDs
print("Unique Employee Numbers : ", hr["EmployeeNumber"].nunique())

# Check unique values in categorical columns
categorical_columns = hr.select_dtypes(include="object").columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(hr[col].value_counts(dropna=False))
```

```python
# Statistical summary
display(hr.describe().T)

# Check number of unique values in each column
for col in hr.columns:
    print(f"{col}: {hr[col].nunique()} unique values")
```

# Dropping Unrelivant Columns

```python
# Remove constant columns from hr
hr.drop(
    columns=["EmployeeCount", "Over18", "StandardHours"],
    inplace=True
)
```

```python
# Verify
print("Updated dataset shape:", hr.shape)
print(hr.columns.tolist())
```

```python
# Check missing values
print("Missing Values : ")
print(hr.isnull().sum().sum())

# Check duplicate rows
print("Duplicate Rows : ", hr.duplicated().sum())

# Check duplicate employee IDs
print("Duplicate Employee Numbers : ",
      hr["EmployeeNumber"].duplicated().sum())
```

## Business-rule validation

```python
print(hr.columns.tolist())
```

```python
# Checking Employee Age
print("Age Below 18 : ", (hr["Age"] < 18).sum())
print("Age Above 65 : ", (hr["Age"] > 60).sum())

# Check Income
print("\nNo Monthly Income : ", (hr["MonthlyIncome"] == 0).sum())
print("Negative Monthly Income : ", (hr["MonthlyIncome"] < 0).sum())

# Checking Work Experience
print("Negative Total Working Years:", (hr["TotalWorkingYears"] < 0).sum())

print("Negative Years at Company:", (hr["YearsAtCompany"] < 0).sum())

print("Years at Company > Total Working Years:",
      (hr["YearsAtCompany"] > hr["TotalWorkingYears"]).sum())

print("Years in Current Role > Years at Company:",
      (hr["YearsInCurrentRole"] > hr["YearsAtCompany"]).sum())

print("Years with Current Manager > Years at Company:",
      (hr["YearsWithCurrManager"] > hr["YearsAtCompany"]).sum())
```

# Export the Dataset

```python
# Saving cleaned dataset
hr.to_csv(
    "hr_employee_attrition_cleaned.csv",
    index=False
)

print("Cleaned dataset exported successfully!")

print("Final dataset shape:", hr.shape)
```
